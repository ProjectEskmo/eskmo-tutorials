# 6. 回報

委託的最新狀態更新會觸發最新 **回報 Reply**，透過回報掌握所有委託最新資訊

回報相關操作，包含：

1. 獲得最新回報
2. 歷史回報查詢
3. 依商品計的委託訊息

------

## 6.1. 最新委託回報

最新的委託回報可以透過 (1) 訂閱 `OrderChanged` 事件得到, 以及 (2) 送出委託後從委託物件中取得

```python
user_id = "A123456789"
password = "*************"
```

```python
from eskmo import api
from eskmo import Stock, User, StockOrder, OrderChangedResult

api.logger.show = False

# 註冊事件
@api.event.order.changed
def onOrderChanged(data: OrderChangedResult):
    # 最新回報
    print(data.order.reply)


api.init()
user: User = api.login(userId=user_id, password=password)

stock: Stock = api.stocks["2888"]
order: StockOrder = stock.order(api.const.ORDER.ACTION.BUY, 11.5, 1).send(user.id)

# 最新回報
print(order.reply)
# 該委託的所有回報
print(order.replys)

print("6.1. Finished!")
api.keepalive() # 為了看到報價, 不讓 Notebook 停止
```

## 6.2. 歷史回報查詢

歷史回報也可以針對指定商品或全商品取得歷史回報

```python
user_id = "A123456789"
password = "*************"
```

```python
from eskmo import api
from eskmo import Stock, User, StockOrder, StockOrders

api.logger.show = False

api.init()
user: User = api.login(userId=user_id, password=password)

# 1. 指定商品的歷史回報查詢
stock: Stock = api.stocks["2888"]
orders: StockOrders = stock.orders(user.id)
for order in orders: 
    print(f"1: {order.replys}")

# 2. 指定序號的歷史回報查詢
all_orders: StockOrders = api.stocks.orders(user.id)
seqNo = "4308201203254"
order: StockOrder = all_orders[seqNo]
print(f"2: {order.replys}")    

# 3. 全部回報查詢
all_orders: StockOrders = api.stocks.orders(user.id)
for order in all_orders: 
    print(f"3: {order.replys}")

print("6.2. Finished!")
api.keepalive() # 為了看到報價, 不讓 Notebook 停止
```

## 6.3. 智慧單委託回報

智慧單委託回報的操作與一般委託相同，以 MIT 為例，訂閱事件為 `MITOrderChanged`，查詢從 `mit_orders` 查詢

```python
user_id = "A123456789"
password = "*************"
```

```python
from eskmo import api
from eskmo import Stock, User, MITStockOrder, MITStockOrders, MITOrderChangedResult

api.logger.show = False

# 註冊事件
@api.event.mit_order.changed
def onMITOrderChanged(data: MITOrderChangedResult):
    # 最新回報
    print(data.order.reply)

api.init()
user: User = api.login(userId=user_id, password=password)

# 1. 指定商品的歷史回報查詢
stock: Stock = api.stocks["2888"]
orders: MITStockOrders = stock.mit_orders(user.id)
for order in orders: 
    print(f"1: {order.replys}")

# 2. 指定序號的歷史回報查詢
all_orders: MITStockOrders = api.stocks.mit_orders(user.id)
seqNo = "13972454"
order: MITStockOrder = all_orders[seqNo]
print(f"2: {order.replys}")    

# 3. 全部回報查詢
all_orders: MITStockOrders = api.stocks.mit_orders(user.id)
for order in all_orders: 
    print(f"3: {order.replys}")

print("6.3. Finished!")
api.keepalive() # 為了看到報價, 不讓 Notebook 停止
```
