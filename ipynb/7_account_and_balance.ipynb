## 7.1. 帳務處理

在初次登入時，會有登入時的帳戶餘額，透過 `user.account.balance` 取得

後續需要透過更新帳戶餘額來獲得更新的餘額

> <br />　
> 請注意，帳戶餘額更新只能每 5 秒執行一次，否則會錯誤
> <br />　

```python
userId = "A123456789"
password = "*************"
```

```python
from eskmo import api

api.logger.show = True

@api.start
def main():
    user = api.login(userId=userId, password=password)

    print(f"Balance: {user.account.balance}") 
    print(f"Summary: {user.account.balance.summary}")
    print(f"Unrealized: {user.account.balance.summary.unrealized}")
    print(f"Updated: {user.account.balance.updated}")

if __name__ == "__main__":
    main()

%reset -f
```
