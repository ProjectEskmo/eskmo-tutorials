## 8.1. 日誌處理

Eskmo 內建提供日誌管理器 `api.logger` 且日誌事件包含 4 個等級：`info`, `warning`, `error`, `fatal`

透過監控 5 個等級的日誌事件來獲得 Eskmo API 日誌相關資訊, 透過 `api.logger.show` 設定 `True`, `False` 來顯示與否

※ 目前預設有無開啟都會自動將每次執行的日誌寫到檔案

```python
from eskmo import api

api.logger.show = False

@api.event.log.info
def onInfoLog(data):
    print("Info log:", data)

@api.event.log.warning
def onWarningLog(data):
    print("Warning log:", data)

@api.event.log.error
def onErrorLog(data):
    print("Error log:", data)

@api.event.log.fatal
def onFatalLog(data):
    print("Fatal log:", data)

@api.start
def main():
    api.logger.info("這是 INFO 消息")
    api.logger.warning("這是 WARN 消息")
    api.logger.error("這是 ERROR 消息")
    api.logger.fatal("這是 FATAL 消息")

if __name__ == "__main__":
    main()    
```

## 4.2. API 執行錯誤

當執行 API 函數遇到錯誤時，將會觸發事件發生

以下模擬帳號密碼登入錯誤，錯誤時可透過對應錯誤事件處理

以登入來說，就是 `@api.event.user.login_fail`

```python
from eskmo import api

api.logger.show = False

@api.event.user.login_fail
def onLoginFail(result):
    print("登入失敗... 原因是:", result)    

@api.start
def main():
    api.login(userId="WRONG_ID", password="WRONG_PASSWORD")

if __name__ == "__main__":
    main()

%reset -f
```

每一個事件的回傳結果都是一個 dataclass，可以透過 `data.[提示]` 獲取事件回傳數據的相關欄位
