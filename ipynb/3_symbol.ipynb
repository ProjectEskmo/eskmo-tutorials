# 3. 商品檔

**商品檔 (Symbol)** 是交易所上特定金融商品的概稱，包含商品基本檔案、交易資訊等。

商品檔包含如商品代碼、漲跌停價等，此類資訊在登入後就會全數準備完成。

除此之外，還有當天會隨時變動的資訊，例如是否緩搓、是否為處置股票等。

> <br /> 
> 商品檔資訊每日 8:15 會進行更新，若登入後長期沒有關閉，需要重新取得商品檔
> <br />

## 3.1. 商品檔資訊

> <br /> 
> 目前商品檔僅支援上市櫃與興櫃的股票 (Stock)
> <br />　

在登入之後，透過 `api.stocks` 可以查詢所有股票相關的商品檔，只要放入指定的商品代號，取得的為 `Stock` 物件

藉由 `stock.info` 了解商品檔資訊，格式為字典

```python
user_id = "A123456789"
password = "*************"
```

```python
from eskmo import api
from eskmo import Stock

api.logger.show = False
api.init()

user = api.login(userId=user_id, password=password)

code = "2330"
stock: Stock = api.stocks[code]
print(stock.info)

%reset -f
```
